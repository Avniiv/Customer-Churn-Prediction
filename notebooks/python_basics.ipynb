```python
customers = [
    {"name": "Avni", "age": 22, "churn": "No"},
    {"name": "Rahul", "age": 25, "churn": "Yes"},
    {"name": "Priya", "age": 30, "churn": "No"}
]
for customer in customers:
    print(customer["name"])
```

```python
for customer in customers:
    print(customer["name"], "-", customer["churn"])
```

```python
def calculate_total(monthly_charge, months):
    total = monthly_charge * months
    return total

result = calculate_total(799, 12)

print(result)
```
